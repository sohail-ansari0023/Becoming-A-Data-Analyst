**step 3: Structural standardaization**<br>
- Strip all the leading trailing spaces from columns headers.
- Transform them into a standard formate either uppercase or lowercase with underscores.
- Remove all of the empty columns from data set used for human convineince.

```python
# Strip all the leading and trailing white spaces form columns headers.
"""
For that we're gonna use 'Vactorised String Approach'.

We must write code that algorithmically targets and destroys whitespaces, regardless
of how many there are. we do this by accessing the .str accessor on the DataFrame's
columns index.
"""

# strip all leading and trailing whitespaces (including types and newlines).
# just changed a column name for testing purpose.
dataframe = dataframe.rename(columns={'First_Name': ' First_Name'})
print(dataframe.columns)
dataframe.columns = dataframe.columns.str.strip()
dataframe.columns.tolist()
```

```python
# 2. Standardizing all columns name into a particular format either upper
# or lower cases with underscore (e.g., FIRST_NAME or first_name)
# we're gonna use the .str accessor for this again

# to convert into upper cases.
# dataframe.columns = dataframe.columns.str.upper() # gengerally it's throwing an error
# AttributeError why it's because there's a culprit in our dataframe columns whose
# type isn't a string, and that's why .str accessor is trying to convert the non-string
# object into str and crashes.

# For that we must force the columns datatypes as string.
dataframe.columns = dataframe.columns.astype(str)

# And than we perform the capitalisation or lower cases.
dataframe.columns = dataframe.columns.str.upper()
# dataframe.columns.tolist()

# To lower cases. with replacing the empty spaces between words to (_) e.g., 'first name' = 'first_name'
dataframe.columns = dataframe.columns.str.lower().str.replace(' ', '_')

dataframe.columns.tolist()
```

```python
"""
Some time non techinal personals include empty columns just for reading and
humnan conveiniences. pandas store these columns as 'unnamed: 0' which is
completely filled with NaN values. we'll completely wipe out such types of columns
via using the dropna() method.
The cool thing about the dropna() method is that it works one both axises,
axis=1 for columns and axis=0 for rows. This is the industry level approach.
"""

# let's drop the completly empty columns we'll it won't be necessary the dataset
# we're working on since it doesn't any columns such like that however it's a
# good practice to achive an industry level data cleaning.
df = df.dropna(axis=1, how='all')
# NOTE: Notice the 'how' argument it state how the droping work. The default is
# 'any' which tells if any of value is NaN, 'all' tells only if all the value is NaN.
```

**step 4: Deduplication**<br>
Handling the duplicated values, extracted data might contains duplicates while extracting from multiples relational sources or via webscraping.

- Identify and drop exact duplicated rows for preventing multi counting.
- Handle partial duplicates via defining the subset keys. like (if a student registered two applications, which one should we keep the older or the latest record)

The proffessional approach for Identifying and eliminating duplicates.

Eact Vs Partial Duplicates:
- _Exact Duplicates_: if every single columns of row A is matching with every single column with B. 
- _Partial Duplicates_: A specific entity(like student or a Lead) appears twice, but some data is different. For examples, a student submitted a form one `tuesdday` and then submit agains on `wednesday`. the student_id would be the same but the timestamp will different.

```python
"""
STEP 1: The Auditing(Never drop blindly).
an Amateurs runs a drop commands and assumes the data is clean. Proffessionals
audit the damage.
"""

# 1. Counts the Exact duplicated rows.
exact_dup_values = dataframe.duplicated().sum()
print(f"Found {exact_dup_values} exact duplicated values.")

# 2. View the actual duplicat rows (keep = False will show the duplicate rows)
dataframe[dataframe.duplicated(keep=False)] # our data set giving the output 0.
# it's because it doesn't contains any exact duplicated values.
```
