### Handling the github crash

#### **Step 1:** Ensure no actual Git process is running
Double-check that you don't have a text editor open (like Vim, VS Code, or Notpad) waitin for you to type a commit message. If you do, save and close it. if not, Proceed to *Step 2*.

#### **Step 2:** Delete the lock file via PowerShell
Since your terminal prompt shows you are using PowerShell (`ps`), run the following command to forcefully remove the stuck lock file:

**PowerShell**

`_Remove-Item -Force .git\index.lock_`

Note: For standard Git Bash, the command would be `rm -f .git/index.lock `.

#### **Step 3:** Verify the fix
Run a simple Git command to ensure everything is workin normally again:

**Bash**

`_git status_`

### Stage and Commit the Changes.

**Step 1: Tell Git to track all changes**
By adding everythin at once, Git is smart enough to look at the deleted files and the new files, realize they are exactly the same, and officially register it as a "rename".

**bash**

`git add .`

**Step 2: Verify Git figured it out (Optional but recommended)**
Run your status command one more time just to be sure git linked the old files to the new folder.

**bash**

`git status`

**Step 3: Commit the changes**
Now, bundle this orgainzation update into a commit.

**bash**

`git commit -m "commit reasons"`

**Step 4: Push to GitHub**
finally, send those updates to your remote repository.

**bash**

`git push`

## **How to Push Remote Repos into the GitHub**

To push a local repository to GitHub, we need to **Initialize Git locally, link it to a newly created GitHub repository, and use the `git push`**

1. **Create a blank repository on github**(Copy the repo URL)

2. **Initialize and Commit My Project Locally**
    - Initialize the local directory as a Git repositor:
        - `git init -b main` (This creates your repository with `main` as the default branch name.)
    - Stage all of my project files for tracking:
        - `git add .`
    - Commit the files to our local repository history:
        - `git commit -m "Initial commit"`
        
3. **Link Local Repo to GitHub and Push**
    - Link my local project to GitHub (replace the URL with the one I copied in Step 1):
        - `git remote add origin https://github.com`
    - Push my local commits up to GitHub:
        - `git push -u origin main` (The `-u` glag sets `main` as my default tracking branch so I can only need to type `git push` for future updates.)

### **To safely remove `Venv` directory from our GitHub repository**

1. **Update our `.gitignore` file**
    - Open the `.gitignore` file in your project. Add `Venv/` on a new line. Save the file(`Ctrl + S`).

2. **Untrack the folder from Git**
    - Open our terminal and run this command to remove the files from Git's memory while keeping them safely on our hard drive:<br>
    `git rm -r --cached Venv/`<br>
    _verfication_: The terminal will output a long list of files being removed(e.g., `rm Venv/...`).

3. **Commit the removal**
    - Save this action to our repository's history by running:<br>
    `git commit -m "Remove Venv directory from tracking"`

4. **Push the fix to GitHub**
    - Update the remote rerpository to clear out the massive file list:<br>
    `git push`<br>
    _Verification_: Refresh our repository page on GitHub in our web browser. The `Venv` folder will no longer be visible.
