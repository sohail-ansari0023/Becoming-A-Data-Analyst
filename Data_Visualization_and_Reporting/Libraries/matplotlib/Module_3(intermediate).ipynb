### **What are Subplots?**

Suppose we're analyzing an e-commerce business.

We want to to see:
- Monthly Sales
- Monthly profits
- Number of customers
- Average order value

We could create four separate charts, but sometimes it is much usful to put these charts together in **one figure**.

That's where **subplots** came in.

> A _Subplot_ is simply an individual plotting area inside a larger figure.

Think:
```text
             ONE FIGURE
┌─────────────────────────────────┐
│                                 │
│   Chart 1       │    Chart 2    │
│                 │               │
│─────────────────┼───────────────│
│   Chart 3       │    Chart 4    │
│                 │               │
└─────────────────────────────────┘
```

Each individual chart is an **Axes**.

The entire container is the **Figure**.

### **The basic `plt.subplots()` function**

We have already used:
```python
fig, ax = plt.subplots()
```
That creates:
```text
1 Figure <----(+)---> 1 Axes
```

Now we can ask Matplotlib for multiple Axes:
```python
fig, axes = plt.subplots(2, 2)
```

This means:
```text
2 rows x 2 columns
```
so we get 4 **Axes** (plotting area).

visually:
```text
┌─────────────┬─────────────┐
│   Axes[0,0] │   Axes[0,1] │
├─────────────┼─────────────┤
│   Axes[1,0] │   Axes[1,1] │
└─────────────┴─────────────┘
```


### **Figure vs Axes becomes even clearer now**

Earlier:
```python
fig, ax = plt.subplots()
```

gave us:
```text
1 Figure + 1 Axes
```

Now:
```python
fig, axes = plt.subplots(2, 2)
```
gives:
```text
Figure
  ↓
 ┌───────┬───────┐
 │ Axes  │ Axes  │
 ├───────┼───────┤
 │ Axes  │ Axes  │
 └───────┴───────┘
 ```

 So the variable name changed now:

 > `ax` to `axes` because we have multiple Axes.

 #### **`figsize` - controlling the overall Figure size**

 Sometimes the default Figure is too small.

 We can control its size:
```python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(10, 6)
)
```

Here:
```text
10 -> Width
6 -> height
```
**The units are inches**

> That means 10 inches wide x 6 inches tall.

#### **`tight_layout()`**

Multiple plots sometimes got:
- titles overlap
- labels cut off
- axes too close together

Matplotlib provides:
```python
plt.tight_layout()
```
Example:
```python
fig, axes = plt.subplots(2, 2, figsize=(10, 6))

# plots....

plt.tight_layout()
plt.show()
```

it automatically adjust spacing between subplots.

### **A proper analyst example**

Let's make a small business performance dashboard.

````python
import matplotlib.pyplot as plt

months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales: list = [20, 25, 30, 35, 42]
profit: list = [5, 7, 8, 10, 12]
customers: list = [100, 120, 140, 160, 190]
orders: list = [50, 60, 75, 85, 100]

fig, axes = plt.subplots(
    2,
    2,
    figsize = (8, 5)
)

axes[0, 0].plot(
    months,
    sales,
    marker='o'
)
axes[0,0].set_title('Monthly Sales')
axes[0,0].set_xlabel('Months')
axes[0,0].set_ylabel('Sales in (000s rs)')

axes[0, 1].plot(
    months,
    profit,
    marker='o'
)
axes[0,1].set_title('Monthly Profit')
axes[0,1].set_xlabel('Months')
axes[0,1].set_ylabel('Profit in (000s rs)')

axes[1, 0].plot(
    months,
    customers,
    marker='o'
)
axes[1,0].set_title('Customers Footfall')
axes[1,0].set_xlabel('Months')
axes[1,0].set_ylabel('Number of Customers')

axes[1, 1].plot(
    months,
    orders,
    marker='o'
)
axes[1,1].set_title('Monthly Orders')
axes[1,1].set_xlabel('Months')
axes[1,1].set_ylabel('Quantity Ordered')

plt.tight_layout()
plt.show()
````

### **Topic 2: Multiple Axes in Matplotlib**

First, let's clear up one possible confusion:

**Multiple Axes != Multiple Subplots**

With subplots:
```text
Figure
│
├── Axes 1
├── Axes 2
├── Axes 3
└── Axes 4
```
We have multiple separate plotting areas.

With multiple axes, we can have **more than one coordinate system associated with the same plotting area.**

For example:
```text
             Sales (₹ lakh)
                  ↑
                  │       ●
                  │     ●
                  │   ●
                  │ ●
──────────────────┼────────────→ Month
                  │
                  │
                  ↓
          Advertising Spend
```
The same plotting region can contain two different y-axes.

### **Why would we need multiple axes?**
Imagine this business data:
```text
Month       Sales       Advertising
Jan         ₹20 lakh    ₹5 lakh
Feb         ₹25 lakh    ₹8 lakh
Mar         ₹30 lakh    ₹10 lakh
Apr         ₹35 lakh    ₹12 lakh
May         ₹42 lakh    ₹15 lakh
```
Now notice:
```text
Sales:
20, 25, 30, 35, 42

Advertising:
5, 8, 10, 12, 15
```
Both are numerical, but they represent **different quantities and units**.

If we put both on one ordinary y-axis:
```python
ax.plot(months, sales)
ax.plot(months, advertising)
```
Matplotlib uses the same y-axis scale.

That can make the smaller series appear visually compressed.

#### **The Solution: `twinx()`**
Matplotlib provides:
```python
ax.twinx()
```
The creates a second y-axis that shares the **same x-axis**.

Basic structure:
```python
fig, ax1 = plt.subplots()

ax2 = ax1.twinx()
```
Now:
```text
             ax1
        Left Y-axis
             ↑
             │
     ┌─────────────────┐
     │                 │
     │     Plot        │
     │                 │
     └─────────────────┘
             │
             ↓
        Right Y-axis
             ax2
```
Both axes share the same x-axis.

#### **Simple example**
Let'e use our previous data.

````python
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales: list = [20, 25, 30, 35, 42]
advertising: list = [5, 8, 10, 12, 15]

# let's start plotting
fig, ax1 = plt.subplots()

ax1.plot(
    months,
    sales,
    marker='o',
    label='Sales'
)

ax2 = ax1.twinx()

ax2.plot(
    months,
    advertising,
    marker='s',
    label='Advertising'
)

plt.show()
````

Now we have:
```text
Left Y-axis  → Sales
Right Y-axis → Advertising
X-axis       → Months
```
This is the fundamental idea behind `twinx()`.

### **Why `twinx()`?**
The name makes sense:
```text
twin
 ↓
two axes sharing something
```
Specifically:
```python
ax2 = ax1.twinx()
```
means:
> **Create another Axes that shares the x-axis with `ax1`.**

So both plots use:
```text
Jan
Feb
Mar
Apr
May
```
on the same horizontal axis.

But they can have different y-axis scales.

#### **Adding proper labels**

Now let's make it professional.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [20, 25, 30, 35, 42]
advertising = [5, 8, 10, 12, 15]

fig, ax1 = plt.subplots(figsize=(9, 6))

ax1.plot(
    months,
    sales,
    marker='o',
    label='Sales',
    color='blue'
)

ax1.set_xlabel('Months')
ax1.set_ylabel('Sales (lakhs)')

ax2 = ax1.twinx()

ax2.plot(
    months,
    advertising,
    marker='s',
    linestyle='--',
    label='Advertising',
    color='green'
)

ax2.set_ylabel('Advertising Expenditure (lakhs)')

ax1.set_title('Sales and Advertising Expenditure')

plt.show()
````

### **Importing warning**

Dual axes might be misleading, since you may found both lines are moving in same directions:

we might conclude, Sale increasing because the advertising expenditure is also increasing.
> "Advertising causes sales to increase."

But our chart only shows that they **move together in this dataset**.

It doesn't establish causation.

There could be:
- Seasonality
- Promotions
- Change in price
- Customer growth
- holidays
- other marketing channels
- external factors

So: 

> **Correlation or visual co-movement does not prove causations**.

This is the same principle we learned in scatter plot.

### **When are multiple axes useful?**

Good examples include:

**Sales + Advertising Spend**
```text
Sales → ₹ lakh
Advertising → ₹ lakh
```

**Revenue + Profit Margin**
```text
Revenue → ₹ crore
Profit Margin → %
```

**Temperature + Rainfall**
```text
Temperature → °C
Rainfall → mm
```

**Website Traffic + Conversion Rate**
```text
Visitors → number
Conversion Rate → %
```
The important feature is:
> **Both varialbes share a meaningful x-axis but have different y-axis scales/units.**

### **Legends with twin axes**

Here's another subtle issue.

If we do:
```python
ax1.plot(months, sales, label='Sales')
ax2.plot(months, advertising, label='Advertising')
```
calling:
```python
ax1.legend()
```
Will generally only show the artists assocaited with `ax1`.

why?

Because `ax1` and `ax2` are separate Axes objects.

**A cleaner approach**

we can retrieve the plotted objects and combine them:

```python
fig, ax1 = plt.subplots()

line1, = ax1.plot(
    months,
    sales,
    marker='o',
    label='Sales'
)

ax2 = ax1.twinx()

line2, = ax2.plot(
    months,
    advertising,
    marker='s',
    label='Advertising'
)

lines = [line1, line2]

labels = [line.get_label() for line in lines]

ax1.legend(lines, labels) 

plt.show()
```

this is a useful pattern when working with multiple axes.

### **Professional analyst rule**

Here's the main thing to remember from this entire topic:

**Use twin axes when:**
- both variables share the same x-axis;
- their units/scales differ;
- showing them together answers a meaningful analytical question.

**Be cautious when:**
- the relationship isn't meaningful;
- the scales are manipulated;
- the chart makes unrelated variables appear related;
- the viewer migh mistake association for causation.

And sometimes, **two seperate subplots are actually clearer than a dual-axis chart**.

## **Modult 3 - Topic 3: Shared Axes**

We've now learned two related but different ideas:

```text
Subplots
-> mulitple seperate plotting areas

Twin axes
-> multiple Axes overlaid in the same plotting area
```

Now we're going to learn **Shared Axes**, which is especially useful when you're comparing mulitple charts.

The key idea is simple:
> **Multiple subplots can share the same x-axis or y-axis scale.**


### **Why do we need shared axes?**
Imagine we're analyzing a company's performance:
```text
Months: Jan, Feb, Mar, Apr, May
```
And we create four subplots:
```text
┌────────────────────┬────────────────────┐
│ Sales              │ Profit             │
│                    │                    │
├────────────────────┼────────────────────┤
│ Customers          │ Orders             │
│                    │                    │
└────────────────────┴────────────────────┘
```
All four charts use the same months.

Without shard axes, each subplot independently manages its x-axis.

But because all four use:
```text
Jan → Feb → Mar → Apr → May
```
We can tell Matplotlib:
> "These subplots use the same x-axis, keep them synchronized."
That's **shared x-axis**.

#### **`sharex=True`**

The syntax is:
```python
fig, axes = plt.subplots(
    2, 
    2, 
    sharex=True
)
```
This means:
```text
2 rows
2 columns
and
all subplots share the x-axis
```
Conceptually:
```text
┌────────────────────┬────────────────────┐
│ Sales              │ Profit             │
│                    │                    │
│                    │                    │
├────────────────────┼────────────────────┤
│ Customers          │ Orders             │
│ Jan Feb Mar Apr May│ Jan Feb Mar Apr May│
└────────────────────┴────────────────────┘
```
Matplotlib synchronizes the x-axis across the relevent Axes.

#### **Let's see it with code**

````python
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales: list = [20, 25, 30, 35, 42]
profit: list = [5, 7, 8, 10, 12]
customers: list = [100, 120, 140, 160, 190]
orders: list = [50, 60, 75, 85, 100]

fig, axes = plt.subplots(
    2,
    2,
    sharex=True,
    figsize=(10, 6)
)

axes[0,0].plot(months, sales)
axes[0,0].set_title("Sales")
axes[0,1].plot(months, profit)
axes[0,1].set_title('Profits')
axes[1,0].plot(months, customers)
axes[1,0].set_title('Customers')
axes[1,1].plot(months, customers)
axes[1,1].set_title('orders')

plt.tight_layout()
plt.show()
````

Now all the subplots uses the same x-axis ctegories.

### **Why is this useful?**

For example:
```text
Sales       → Jan → Feb → Mar → Apr → May
Profit      → Jan → Feb → Mar → Apr → May
Customers   → Jan → Feb → Mar → Apr → May
Orders      → Jan → Feb → Mar → Apr → May
```
The x-axis represents **time** for every chart.

Sharing it helps the viewer visually align the information.

we can look vertically:
```text
             Jan     Feb     Mar     Apr     May

Sales         ●──────●──────●──────●──────●

Profit        ●──────●──────●──────●──────●

Customers     ●──────●──────●──────●──────●

Orders        ●──────●──────●──────●──────●
```
Now it's much easier to ask:
> What happened around March?

We can visually move down the charts and inspect March across all metrics.

### **What does `shrey=True` do?**

Exactly the opposite concept.
```text
fig, axes = plt.subplots(
    2,
    2,
    sharey=True
)
```
It tells matplotlib:
> "These subplots should share the same y-axis scale."

Suppose we're comparing sales between different regions:
```text
North
South
East
West
```
And each chart represents sales over months.

If the y-axis represents:
```text
0 -> ₹50 lakh
```
for all of them, sharing the y-axis can make the comparisons much easier.

#### **Why Shared Y-axis matters**
Imagine:
```text
Region A:
sales = 10, 20, 30, 40, 50

Region B:
sales = 5, 10, 15, 20, 25
```
if matplotlib plot them in different subplots and assign different y-scale to both, The visual hights can sometimes shows he same pattern to make look them same , however the values are different.

The data is showing Region A has more potential sale as compared to Region B, but the chart won't help us to tell the difference.

### **Example: Regional comprison**
Let's create four regional sales charts.

````python
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

north: list = [20, 25, 30, 35, 40]
south: list = [15, 20, 24, 26, 30]
east: list = [10, 18, 28, 35, 45]
west: list = [25, 30, 40, 50, 55]

fig, axes = plt.subplots(
    2,
    2,
    sharex=True,
    sharey=True,
    figsize=(10, 6)
)

axes[0, 0].plot(months, north)
axes[0, 0].set_title('North Sales')

axes[0, 1].plot(months, south)
axes[0, 1].set_title('South Sales')

axes[1, 0].plot(months, east)
axes[1, 0].set_title('East Sales')

axes[1, 1].plot(months, west)
axes[1, 1].set_title('West Sales')

plt.tight_layout()
plt.show()
````

Now both dimensions are synchronized:
```text
X-axis -> Same months
Y-axis -> Same sales scale
```

### **We can share only within rows and columns**

We don't always need every plots to share the same axis.

Matplotlib supports:
```python
sharex='row'
sharex='col'
```

This basically means share the axis on plots in the same row or same column.

### **Shared axes aren't only about appearance**
This is a deeper point.

Suppose we're comparing:
```text
Product A
Product B
Product C
```
and all three charts represent:
```text
Montly Sales
```
Using the same y-scale means:
> A vertical distance has the same numerical meaning across charts.

That's extremely important for honest comparison.

**Without Shared Y-axis**

One chart might automatically use: 0 -> 100

while another uses: 0 -> 20

> Their lines would look similar tall even thought the actual values differ greatly.

**With Shared Y-axis**

both might use: 0 -> 100

> Now the visual comparison is much more meaningful.

### **When sould we NOT share an axis?**

Suppose we have:
```text
Chart A:
Revenue -> ₹ crore

Chart B:
Customer Age -> years
```
There is no meaningful reason to force them to share the same y-scale.

Because the variables have completely different scale and meanings.

So:

> **Share axes when sharing improves comparison---not just because matplotlib allows it.**

### **The big picture so far**

we've now learned three things that sound similar but are actually different:
```text
1. Subplots
   Multiple plotting area

2. Twin Axes
   Multiple axes overlaid in one plotting area.

3. Shared Axes
   Multiple plotting areas with synchronized axis scales
```

This distinction is worth memorizing conceptually.

## **Module 3 — Topic 4: Annotations**

So far, our charts have mostly answered:
> **"What does the data look like?"**

Annotations let us go one step further:
> **"What important thing should the viewer notic?"**

For a data analyst, that's a big difference.

### **1. What is an annotation?**

An annotation is additional explonatory information placed directly on a chart.

It can be:
- text
- an arrow
- a highlighting point
- a label
- an explanation of an unsual event

For example:
```text
Sales
 ↑
40│                    ● ← Highest sales
35│                ●
30│            ●
25│        ●
20│    ●
  └──────────────────────→
```

### **2. The basic `annotate()` function**

Matplotlib provides:
```python
ax.annotate()
```
The simples exampl:

````python
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales: list = [20, 25, 30, 35, 42]

# plotting
fig, ax = plt.subplots()

ax.plot(months, sales, marker='o')

ax.annotate(
    'Highest Sales',
    xy=('May', 42)
)

ax.set_title('Monthly Sales')
ax.set_xlabel('Months')
ax.set_ylabel('Sales (₹ lakhs)')

plt.show()
````

Here:
```python
xy=('May', 42)
```
tells matplotlib:
> "Put the annotation at the data point corressponding to May and 42."

### **3. But where should the text appear**

In the previous example the text appear exactly at the point ('May', 42) which basically overwrite at the data point.

This is where:
```python
xytext
```
comes in.

Consider:
```python
ax.annotation(
    'Highest sales',
    xy=('May', 42),
    xytext=('Mar', 45)
)
```
Now:
```text
xy -> actual data point

xytext -> Where the annotations text should appear
```
Conceptually:
```text
Sales
 ↑
45│              Highest sales
  │                    ↑
42│                    ●
  │                  May
  │
  └────────────────────────→
       Mar          May
```

So:
```python
xy=('May', 42)
```
= **Point we're talking about**

While:
```python
xytext=('May', 45)
```
= **location of the annotation text**

This distinction is fundamental.

### **4. Add an arrow**
Now we can connect the text to the point:
```python
ax.annotate(
    'Highest Sales',
    xy=('May', 42),
    xytext=('May', 45),
    arrowprops=dict(arrowstyle='->')
)
```
Now the chart communicates:
```text
                 Highest sales
                      |
                      ↓
                     ●
                    May
```
The arrow makes the relationship obvious.

#### **The `arrowprops`**

It's means of usage is to control the arrow.

which mean:
> Draw an arrow pointing towards the annotated data point.

we can also control things like:
```python
arrowprops=dict(
    arrowstyle='->',
    connectionstyle='arc3'
)
```
Well for now don't worry about the arrow option yet. The important thing right now is understanding the structure.

### **Complete Example**
Let's make a proper analyst-style chart.

````python
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales: list = [20, 25, 35, 40, 45]

fig, ax = plt.subplots(
    figsize=(8, 5)
)

ax.plot(
    months,
    sales,
    marker='o'
)

ax.annotate(
    'Heighest sales',
    xy=('May', 45),
    xytext=('Mar', 44.5),
    arrowprops=dict(
        arrowstyle='->'
    )
)

ax.set_title('Monthly Sales')
ax.set_xlabel('Months')
ax.set_ylabel('Sales')

plt.tight_layout()
plt.show()
````

Notice the logic:
```text
Plot data -> Identify important points -> Specify point xy -> Specify text position with xytext -> Connect them with arrowprops
```

### **5 Annotating an unusual value**

Suppose we have:
```python
sales = [20, 25, 30, 35, 60]
```
May suddenly jumps to ₹60 lakh.

We might annotate:
```python
ax.annotate(
    'Unexpected sales spike',
    xy=('May', 60),
    xytext=('Mar', 60),
    arrowprops=dict(
        arrowstyle='->',
    )
)
```
Now the viewer immediately knows:
> Something unusual happened in May

But here's an important analyst principle:

**The annotation should not claim something the data doesn't prove.**

For example:
```text
"Advertising caused the sales spike"
```
Just because advertising also increased.

Instead:
```text
"Sales spike in May"
```
or, if we have documented business context:
```text
"Sales spike following May compaign."
```
That distinction connects directly about **correlation vs causation**.

#### **We can annotate multiple points inside a single chart**

That doesn't mean we annotate every points.

> **Annotate the important things, not everything.**

### **A professional analyst mindset**

when looking at a chart, ask:
```text
1. Is there any important peak?
2. Is there any unsual drop?
3. Is there an outlier?
4. Is there a business event?
5. Is there meaningful change?
6. Is there something the audience might miss?
```
If yes -> annotation may help.

If no -> don't force one.
