#### **Our First Practical Dataset**

We'll use business-oriented datasets throughout the course.

Let's create a simple dataset:

````python
import pandas as pd
import matplotlib.pyplot as plt

data: dict = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [120000, 135000, 128000, 160000, 175000, 190000],
    'Profit': [20000, 24000, 21000, 30000, 35000, 42000],
}

df: pd.DataFrame = pd.DataFrame(data=data)

df
````

#### **Our First Matplotlib charts**.

````python
# we have already imported matplotlib.pyplot as plt.
plt.plot(df['Month'], df['Sales'])
plt.show()
````

````python
plt.plot(df['Month'], df['Sales'])

plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
````

Now the graphs communicate much more clearly....

#### **Add markers**

The markers place a visible point at each observations.

````python
plt.plot(
    df['Month'],
    df['Sales'],
    marker='o' # This puts 'o' at every observations.
)

plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
````

#### **Add a Grid**

````python
plt.plot(
    df['Month'],
    df['Sales'],
    marker='o'
)

plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(True)

plt.show()
````

### **Mdule 1 — Matplotlib Architecture**

**1. What exactly is Matplotlb?**

When we create a graph, there isn't just "a graph."

There are objects representing:
- the entire canvas
- plotting areas
- x/y axes
- lines
- points
- text
- titles
- legends
- ticks
- grids

Matplotlib manages these objects and eventually renders them as an image on our screen or into a file.

The deeper idea is:
> Matplotlib provides a system of objects that represent everything we see in chart.

**2. The most important concept: Figure vs Axes**

Consider this:
```python
fig, ax = plt.subplots()
```
we have created two things.
```text
fig -> Figure
ax -> Axes
```
let's understand them separately.

#### **What is a Figure?**

A figure is the entire container/canvas on which our visualization is drawn. Think of it as the **Whole sheet of paper**.
```text
┌─────────────────────────────────────┐
│             FIGURE                  │
│                                     │
│      ┌─────────────────────┐        │
│      │                     │        │
│      │       AXES          │        │
│      │                     │        │
│      │      📈             │        │
│      │                     │        │
│      └─────────────────────┘        │
│                                     │
└─────────────────────────────────────┘
```
A figure can contain one or multile Axes.

````python
# Example
fig = plt.figure()
plt.show()
````

#### **What is Axes?**

An **Axes** is the actual plotting area where data is visualized.

For example:
```text
Figure
┌──────────────────────────────┐
│                              │
│    ┌────────────────────┐    │
│    │                    │    │
│    │     DATA PLOT      │    │
│    │                    │    │
│    └────────────────────┘    │
│                              │
└──────────────────────────────┘
```
The inner rectangle is the **Axes**.

And here's the confusing part:
> Matplotlib calles the plottin area `Axes` not `Axis`.

Technically, an `Axes` object contains two `Axis` objects:

```text
Axes
 ├── XAxis
 └── YAxis
```

#### **Create Figure + Axes**
Now let's write professional Matplotlib code.

````python
fig, ax = plt.subplots()

plt.show()
````

#### **Why Use `ax`?**
Now we can directly tell the Axes what to do.

````python
fig, ax = plt.subplots()

ax.plot(
    [1, 2, 3, 4],
    [10, 20, 15, 30]
)

plt.show()
````

Notic: 
```python
ax.plot()
```

rather than:
```python
plt.plot()
```

This is called object-oriented interface.

#### **The Object-Oriented Mental Model**

Instead of thinkin:
>"I'm Calling random matplotlib functions."

Think:
> "I'm Modifying objects."

````python
# A complete example:
months: list = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales: list = [120, 135, 128, 160, 175, 190]

fig, ax = plt.subplots()

ax.plot(months, sales)

ax.set_title('Monthly Sales')
ax.set_xlabel('Months')
ax.set_ylabel('Sales (in units)')

plt.show()
````

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [120, 135, 128, 160]
profit = [20, 24, 21, 30]

fig, axes = plt.subplots(1, 2)

axes[0].plot(months, sales)
axes[0].set_title("Sales")

axes[1].plot(months, profit)
axes[1].set_title("Profit")

plt.show()
````

#### **Two styles of matplotlib**
Matplotlib essentially gives us two common ways to work.

**Style 1 - Pyplot interface**

```Python
plt.plot(x, y)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
```
Simple and convenient.

**Style 2 - Object-Oriented interface**

```python
fig, axes = plt.subplots()

axes.plot(x, y)
axes.set_title('Monthly Sales')
axes.set_xlabel('Months')
axes.set_ylabel('Sales')

plt.show()
```
More explicit and easier to manage for complex visualizations.

#### **Which one should a Data Analyst prefer?**

For very simplr plts:
```python
plt.plot(...)
```
is perfectly fine.

But as our visualization become complex, I recommend developing an **Object-Oriented interface** matplotlib habit:
```python 
fig, axes = plt.subplots(...)
```

### **Part A —— Line Plot**

A **line plot** connects data points with lines to show how anmerical variable changes, especially across an ordered variable such as time.

The most common use:
> **Trend Analysis**

#### **Why does it exist?**

Humans are very good at recognizing movement.

Compare:
```text
Jan  100
Feb  120
Mar  135
Apr  150
May  170
```
With:
```text
100 ──●
       ╲
120     ●
         ╲
135       ●
           ╲
150         ●
             ╲
170           ●
```

The second representation makes the **direction and rate of change** easier to see.

#### **Basic Line Plot**
Let's work on previous business data.

````python
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'Jun']
sales = [120, 128, 121, 145, 160, 190]

# let's create the figure and axes object.
fig, axes = plt.subplots()

# plot the chart. / marker attribute will mark the indvidual points
axes.plot(
    months, # x-axis variable
    sales,  # y-axis variable
    marker='o',
    linestyle= '--', # we can even control the appearance of the line
    ) 
axes.set_title('Monthly Sales') # Setting the chart title.
axes.set_xlabel('Months ————>') # Setting the label of x-axis.
axes.set_ylabel('Sales in (000s) —————>') # Setting the label of y-axis.
axes.grid() # making the grid

# We can control the appearance of the line too.


plt.show() # the final step.
````

Now the chart communicates what the variables mean.

The "`o`" creates circular markers.

some common line styles:
```text
"-"      Solid
"--"     Dashed
":"      dotted
"-."     dash-dot
```
Do not use different line styles only for decoration.

Use only them when they help distinguish meaninful series.

#### **Multiple Lines**

Now the company gives us another variable "profit". 

We can plot both of them sales and profit into the same axes.

````python
# let's copy paste the other variables.
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'Jun']
sales = [120, 128, 121, 145, 160, 190]
profit = [20, 25, 22, 30, 35, 42] # Added the new variable.

fig, ax = plt.subplots()

ax.plot(
    months,
    sales,
    marker = 'o',
    label= 'Sales',
    linestyle = '-'
)

ax.plot(
    months,
    profit,
    marker='o',
    label= 'Profit',
    linestyle = '--'
)

ax.set_title('Monthly Sales and Profit')
ax.set_xlabel('Months')
ax.set_ylabel('Amount')

# add the legend: The legen says which line represents which variables.
ax.legend() 

plt.show()
````

#### **Important analytical problem**

Look carefully.
```text
Sales: goes from 120 -> 180

profit: goes from 20 -> 42
```

Both are measured in different scales.

Plotting them on the same Y-axis may makes one series appears visually smaller.

This is where we use **Secodary axes/ Twin axes**.

often, **separate charts or normalized values communicate the comparison better.

#### **When should we use Line chart?**

Excellent use cases:
- Sales over time (Month -> Sales).
- stock prices (Date -> price).
- Website traffic (Day -> Visitors).
- Monthly profit (Month -> profit).
- Temperature (Date -> Temperature).

The commone patter is:
> **X has a meanigful order**.

**Analyst Rule**
> Don't connect points just because you can.

A line implies continuity or meaningful order.

### **Part B —— Scatter Plot**

Now we move to one of the most important charts for EDA.

**Difinition**

A _Scatter plot_ displays individual observations using points to examine the relationship between two numerical variables.

For example:
```text
Advertising spend -> Sales
```

#### **Why does scatter plot exist?**

Imagine:
```text
Advertising     sales

10               100
20               120
30               140
40               155
50               180
```

We might suspect:
> Higher advertising expenditure is associated with higher sales.

A Scatter plot lets us visually investigete the relationship.

#### **Basic Scatter plot**

````python
advertising: list = [10, 20, 30, 40, 50, 60, 70]
sales: list = [100, 115, 130, 150, 155, 175, 190]

fig, ax = plt.subplots()

ax.scatter(
    advertising,
    sales,
    alpha=0.5
)

ax.set_title('Advertising vs Sales')
ax.set_xlabel('Advertising spend in (000s) ---->')
ax.set_ylabel('Sales in (Units) ---->')

plt.show()
````

Each point represents an observation.

For example:
```text
Advertising = 30
Sales = 130
```
becomes one point.

#### **How to interpret a scatter plot**

This is more important than syntax.

Imagine the points look like:
```text
Sales
 │
 │               ●
 │           ●
 │        ●
 │      ●
 │   ●
 │ ●
 └────────────────────
       Advertising
```
That's a **positive relationship**.

As advertising increases, the sales increase too.

**Negative relationship**

```text
Y
│ ●
│    ●
│       ●
│          ●
│             ●
└──────────────── X
```
As X increases, Y tends to decrease.

**No Obvious Pattern**

```text
Y
│     ●     ●
│ ●       ●
│       ●
│   ●       ●
│      ●
└──────────────── X
```
No clear patten.

#### **Correlation vs causation**

This is **extremely** important for analyst

Suppose we find:
```text
Advertising ↑
Sales ↑
```

we cannot automatically assume:
> Advertising causing sales to increase.

Maybe:
- seasonal demand increased.
- a new product launched.
- prices changed.
- competitors had stock problem.
- marketing and sales were both affected by another factor.

So:
> Correlation does not automatically establish causation.

A scatter plot shows a relationship/pattern.

It doesn't by itself prove why that relationship exists.

#### **Add transparency**

Sometimes we have thousands of observations.

Points overlap.

we can use:

```python
ax.scatter(
      months,
      sales,
      alpha= 0.5
)
```

`alpha` controls transparency.

For example:
```text
alpha = 1.0 -> fully opaque
alpha = 0.5 -> semi-transparent
alpha = 0.1 -> very transparent
```
This can help reveal dense regions.

### **Scatter plot with categories**

Suppose we have:

### **Part C —— Bar Chart**

Now we move to categorical comparison.

**Difinition**

A _Bar Chart_ represents values for different categories using recatangular bars.

Example:
```text
Department   Revenue

Electronics  900
Clothing     700
Furniture    500
Grocery      800
```

Bar chart:
```text
Electronics █████████
Clothing    ███████
Furniture   █████
Grocery     ████████
```

Instant comparison.

#### **Why does a bar chart exists?**

The human eyes are very good at comparing lengths. so bar chart make category differences easy to see.

**Basice Bar Chart**

````python
regions: list = ['North', 'South', 'East', 'West']
revenue: list = [500, 700, 400, 900]

fig, ax = plt.subplots() # create the object

ax.bar(
    regions,
    revenue
)

ax.set_title('Revenue by Regions')
ax.set_xlabel('Region')
ax.set_ylabel('Revenues in (000s)')

plt.show()
````

#### **Horizontal bar chart**

Sometimes category names are long. so,

Instead of:
```python
ax.bar(...)
```
use:
```python
ax.barh(...)
```

**Example:**

````python
products = [
    'Laptop',
    'Smartphone',
    'Wireless Headphones',
    'Smart Television'
]

sales = [500, 400, 550, 350]

fig, ax = plt.subplots()

bars = ax.barh(
    products,
    sales,
)

ax.set_title('Product Sales')
ax.set_xlabel('Product category')
ax.set_ylabel('Sales in (units)')

# A professional chart may benefit from displaying the values:
ax.bar_label(bars)

plt.show()
````

Horizontal bars are often easier to read when category labels are long.

#### **Bar chart vs Histogram**
