# **How To Collect Data from an Excel Sheets/Workbooks**

To Extract Data from a relational Mutli-sheets Workbooks in a proper, industry standard manner.
we must treat our Python scripts as an automated Databse query.

An efficient will not rely on opening every files manually, instead read the whole workbook into the memory, isolates the specific tables, and the perform relational algebra to merge them together in an efficient way.

let's get a specific blue print for extracting and merging tables from `Data_Collection_and_Extraction/Via_Spreadsheets/Spreadsheets/Multisheets_Universtiy_DBMS.xlsx` in industry standard manner.

**Workbook Details:**
- Contains three sheets.

    - `Students`
    - `Courses`
    - `Grades`





### **Phase 1: Ingestion layor**
Instead of reading the file three separate times(which waste I\O processing). we read it once and capture all the sheets.

```python

# import the required libraries
import pandas as pd
import os

# Hardcoding the file path.
file_path: str = 'Spreadsheets/Multisheets_University_DBMS.xlsx'

# Always insure the file exists.
if not os.path.exists(path=file_path):
    raise FileNotFoundError(f'Error! File not Found {file_path}')

# Opening the entire workbook as a dictionary via setting (sheet_name=None).
workbook: dict = pd.read_excel(io=file_path, sheet_name=None)

# Isolating the Sheets.
df_student = workbook['Students']
df_courses = workbook['Course']
df_grades = workbook['Grades']

print(f'Extraction Successfull! Loaded number of sheets: {len(workbook)}')
```

### **Phase 2: The realational merges(Joins)**

let's first leftjoin our Student and Grades DataFrames on `Student_Id` and make a new `report_card_df`.

And after that we will left join our `report_card_df` to our courses df , by `Course_ID`

**Note:** while performing merges , strictly watch the columns name (as it is in previous DataFrames) to prevent the pipeline crashing.


```python
# First Merge Student table to Grades with Student_ID
report_card_df: pd.DataFrame = pd.merge(
    left=df_student,
    right=df_grades,
    how='left',
    on='Student_ID', 
)

# Second merge with report_card_df to Course table on Course_ID
final_report_card: pd.DataFrame = pd.merge(
    left=report_card_df,
    right=df_courses,
    on='Course_ID',
    how='left'
)
```
