## **Overview of the Database Design Process**

### **1. What is Database Design?**

Database design is the systematic process of deciding:
- What data needs to be stored.
- How that data should be organized.
- How different piece of data are related.
- What rules/constraints should apply.
- How the design will eventually be converted into database tables.

In simple words:
> **Database design is the process of creating logical blueprint for a database before actually builiding it.**

### **2. Why do we need Database Design?**

Suppose a college says:
> "We need a database to store student information."

A beginner might immediately create:
```text
Student
---------------------------------------------------------
ID | Name | Course | Teacher | Department | Fees | ...
```
It looks simple.

But now imagine:
- One student takes mulitple courses.
- One teacher teaches multiple students.
- One department has many teachers.
- Fees may change.
- A student can change departments.
- A course can have many students.

Suddenly, the single table becomes messy.

We may get:
- duplicated data
- inconsistent data
- unnecessary storage
- difficult updates
- insertion problems
- deletion problems
- confusing relationships

So database design exists to prevent these problems **before implementation.**

### **3. The Database Design Process**

A typical database design process can be understood as:
```text
Real-world requirements
          ↓
Requirements analysis
          ↓
Conceptual Design
          ↓
ER model/ ER diagram
          ↓
Logical Design
          ↓
Relational Schemas
          ↓
Normalization / constraints
          ↓
Physical Design
          ↓
Implementation
          ↓
Testing & maintenance
```
Let's understand each stage.


### **Table to remember**

| **Concept** | **Meaning** |
|-------------|-------------|
|Database Design | Planning the structure of a database |
| Requirement Analysis | Understanding what users/Business need |
| Conceptual Design | High-level representation of data and relationships |
| E-R Model | Model used to represent entities, attributes and relationships |
| Logical Design | Convert conceptual model into logical database structures |
| Schema Refinement | Imporve design and reduce redundancy/problems |
| Physical Design | Decide efficient storage/acceess mechanisms |
| Implementation | Create the actual Database |
| Testing | Verify correctness and constraints |
| Maintenance | Modify and imporve database as requiremnets evolve |

The **big picture** to keep in our head:

> **Database design is not "creating tables". Creating tables is only one later step.**

The real process is:
> **Understand the real world -> model it -> refine it -> convert it -> implement it -> test it -> maintain it.**

### **Part 2 — The E-R Model**

This is one of the most important parts of database design because the E-R model is the bridge between the real-world problem and the relational database.

#### **1. What is the E-R Model?**

E-R Stands for _Entity-Relationship._ It was introduced by **Peter Chen** 1976 and became one of the fundamental approaches to conceptual database design.

The **Entity-Relationship model** is a conceptual data model used to represent:
1. What objects/entities exist in a system. (Relation Name)
2. What information describes those entities. (Attributes)
3. How those entities are related to one another. (Relationships)

In very simple terms:

> **The E-R model describes the important things in the real world and the relationshps between those things.**

#### **2. Why do we need the E-R Model?**

Why not directly the SQL tables?

Because real-world requirements are generally expressed via the natural language.

For Example:
> "Each student shoud have unique Id and can take part on several courses, Every course belongs to a department."

A database developer has to convert that statement into a precise structure.

The E-R model gives us an intermediate conceptual layer:

```text
Business Requirements -> E-R model -> Relational Tables -> SQL
```

Without this conceptual layer, database design can become confusing very quickly.

#### **3. The Three Fundamental Components**

The E-R model primarily revolves around three ideas:
```text
ENTITY
ATTRIBUTE
RELATIONSHIP
```
Think of them as:
> Thing -> Information about the thing -> Connection between thigs

For example:

```text
STUDENT(Student_ID, Name, Age)

STUDENT <-----Enrolls in ------> COURSES
```
So:
- `STUDENT` = Entity
- `Student_ID`, `Name`, `Age` = Attributes
- `Enrolls in` = Relationship.

Let's quickly understand them.

**Entity**: An entity is distinguishable real-world object, person, place, thing, event, or concept *about which we want to store information.*

Examples:
```text
Student
Course
Teacher
Department
Employee
Customer
Product
Bank Account
Order
```
_Entity set_( An collection of similar entities of the same type.)

**Attributes**: An Attribute is the propery or charecterstics that describes an entity.

For `STUDENT`:
```text
Student_ID
Name
Age
Email
Phone
```
are attributes.

For `Product`:
```text
Product_ID
Product_name
Product_category
Product_price
```
are attributes.

### **Types of Attributs**

Attributes can be classified into several types:
```text
Simple
Composite
Single-Valued
Multivalued
Derived
Stored
Null/Optional
key attributes
```
Let's tackel them one-by-one.

#### **1. Simple Attribute**

A _simple attributes_ are those which cannot be meaningfully divided into smaller components/attributes for the application purposes.

Example:
- Age
- Gender
- Salary
- Student_ID

```text
Age = 21
```
We normally don't divide `Age` into smaller meanigful attributes.

#### **2. Composite Attributes**

A _composite attributes_ are those who **can** be divided into smaller meaningful components/attributes.

Example:
```text
Name - First -> Middle -> Last
Addresses - City -> State -> District
```

### **Simple vs Composite Attribute**

| **Simple** | **Composite**|
|------------|--------------|
| Cannot be meaningfully divided | Can be divided into componenets|
| Age | Full Name |
| Gender | Addressess |



### **Single-valued Attribute**

A **Single-valued** attribute has one value for a particular entity at a given time.

Example:
```text
Student_ID = 101
Date_of_birth = 08/03/2003
```
A student has one Student_ID, a person normally has only one Date of birth.

So: 
```text
Student
    |
    --> DOB = one value
```
### **Multivalued Attribute**

A **multivalued attribute** can have multiple values for one entity.

Example:
- A student might have multiple phone number.
- An Employee can multiple sets of Skills (python, Power BI, DBMS).

These are multivalued attribute.


### **Stored Attribute**

A **Stored Attributes** are those those attribute which are directly stored in the database.

Example:
- Date of birth
- Unit price
- Quantity

### **Derived Attribute**

A **derived attributes** are those whose value can be calculated from other stored data.

Example:
- Age -> Can be calculated by DOB.
- Total_purchase -> Can be calculated by multiplying Unit price and Quantity.
- Delivery Status -> Can be provided by the help of Delivery Date.

#### **Stored vs Derived**

| **Stored** | **Derived** |
|------------|-------------|
| Directly Stored | Obtain by the help of stored values |
| DOB | AGE |
| Unit Price | Total purchase |

**Why this is important**

> Because storing derived values might consume unnecessary memory, and can create inconsistency.

### **Key Attribute**

A **key attribute** are those which helps to uniquely identify an entity.

For students:
- Student_ID

For employees:
- Employee_ID

For products:
- Product_ID

For banks:
- Account_Number

### **NULL/Optional Attribute**

Sometimes an attribute may not have a value for a particular entity.

For example:
```text
Middle_Name
```
A student may not have one.

Or:
```text
Apartment_Number
```
may not apply for someone whose living in individual houses.

So an attribute can sometimes be optional.

### **Relationship**

Enteties does not exist independently in a database.

They interact.

For example:
```text
STUDENT -------Enrolls in ------- COURSE
```

Tha phrase:
> "Students enrolss in course"

describes a **relationship**.

Therefore:
> **A relationship represents an association between entities**.

Examples:
```text
CUSTOMER -----Places-----ORDER

EMPLOYEE -----Works for--------DEPARTMENT

DOCTOR --------Treats--------Patients

TEACHER --------Teaches----------COURSE
```

### **Degree of a Relationship**

The degree of relationship refers to the number of entity types participating in the relationship.

The most common types are:
```text
Unary
Binary
Ternary
N-ary
```

#### **Unary Relaionships**

When the relationships forms between only one type of entity.

Example:
```text
Employee -------Supervises--------Employee
```
An employee supervises an employee.

This is an unary/recursive relationship.

#### **Binary Relationship**

When the relationships froms between two types of entity.

Example:
```text
STUDENT --------Enrolls in ------- COURSE
```

There are two entities `STUDENT` And `COURSE`.

This is extremely common.

#### **Ternary Rlationship**

The relationship forms between Three entities.

Example:
```text
SUPPLIER
PRODUCT
PROJECT
```

> A supplier sends the product for a project.

#### **N-ary relationship**

Relationships forms between more than **Three entities**.

Example:
```text
ENTITY A
ENTITY B
ENTITY C
ENTITY D
      ↓
  RELATIONSHIP
```

### **Cardinality in relationships**

**Cardinality** tells us how many of entities of one type can or may be associated with entities of another type.

Consider:
```text
STUDENT --------Enorlls in ---------COURSE
```
> How many students can be enrolled in one Course?
Maybe:
```text
many
```

> How many courses can a student enrolls?
Also:
```text
many
```

This is:

**Many-to-Many(M:N)**

#### **common cardinality types**
- Many-to-Many (M:N)
- Many-to-one(N:1)
- One-to-Many(1:M)
- One-to-One(1:1)
