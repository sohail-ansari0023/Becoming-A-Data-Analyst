## **Removing Redundant Attributes in Entity Sets**

> An entity has attributes, but **not every piece of information related to an entity shouldd necessarily be stored as its attribute.**

### **What is Redundancy?**

In database design, _redundancy_ means storing the same information unnecessarily in more than one places.

For example, suppose we have:
```text
STUDENT
-----------------------------
Student_ID
Name
Course_ID
Course_Name
Deprtment_Name
```
imagine 500 students are enrolled in the same course:
```text
Student_ID | Name   | Course_ID | Course_Name | Department
-----------|--------|-----------|-------------|------------
101        | Rahul  | C101      | DBMS        | Commerce
102        | Aisha  | C101      | DBMS        | Commerce
103        | Ahmed  | C101      | DBMS        | Commerce
104        | Sara   | C101      | DBMS        | Commerce
...
```
Notice what happened.

`Course_Name = DBMS` is repeated for every student taking that course.

`Department = Commerce` is also repeated.

The same fact are being stored again and again.

That's redundancy.

### **Why is Redundancy a Problem?**

At first, duplicate data may not look Dangerous. But it creates several problems.

The major ones are:
- Storage wastage
- Insertion Anomaly
- Update Anomaly
- Deletion Anomaly
- Inconsistency
- Maintenance difficulty

Let's understand them....
