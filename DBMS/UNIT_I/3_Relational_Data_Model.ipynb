## **Relational Data Model: Structure of relational databases**

#### **1. What is a Data Model?**

A _data model_ is conceptual framework that describes:
- How the data is organized,

- How different data items are related,

- What rules apply to the data, and

- How users can work with that data.

In simple words:
> A data model is blueprint of **Organizing data**.

#### **2. What is a Relational Data Model?**

A **Relational Data Model** is a data model in which data is represented using **relations**, which are commonly known as **tables** consisting or rows and columns.

It was formally proposed by **Edgar F. Codd** in 1970.

The central Ideas is surprisingly simple.
```text
Data
 ↓
Relations
 ↓
Tables
 ↓
Rows + Columns
```

But underneath this simple table structure is a mathematical foundation based on **set theory and relational algebra**.

#### **3. Why was the Relational Data Model Created?**

Before the relational model became dominant, databases could be represented using models such as:
- Hierarchical model
- Network model

These could represent relationships, but designing and querying complex data could become difficult.

Codd's relational approach introduced much cleaner idea:
> Store data in simple relation and use mathematical operations to manipulate those relations.

For example, instead of thinking:
```TEXT
Studen -> Course -> Teacher
```
as a complicated navigational structure, we can represent them as separate relations:
```text
STUDENT
COURSE
TEACHER
```
and establish connections using keys.

This made database systems more flexible and easier to query.

#### **4. Basic Structure of Relational Database**

**Relation**
> A relaion is a collection of tuples having the same set of attributes.

In practical database language:
> Relation ≈ Table

For example:
| **Student_ID** | **Name** | **Course** |
|----------------|----------|------------|
| S101 | Sohail | B.Com |
| S102 | Rohit | BBA |
| S103 | Ayesha | B.Com |

This table can be represented formally as:
```text
STUDENT(Student_ID, Name, Course)
```
Here:
```text
STUDENT = Relation
Srudent_ID = Attribute
Name = Attribute
Course = Attribute
```

**Important Note**
In formal relational theory, a relation is more than simply "a table on screen". It has mathematiacal properties. But for most introductory DBMS work, thinking **relation = table** is a useful starting point.

**Tuple**

A tuple is a single record/row in a rlation.

```text 
101 | Sohail | B.Com |
```
is one tuple.

**Attribute**

An _Attribute_ is a named property or charecteristics of an entity represented as a column in a relation.

For example:
| **Student_ID** | **Name** | **Course** |
|----------------|----------|------------|

Therefore:
> **Attribute** = **Column**

**Domain**

A _Domain_ is set of valid values that an attribute can take.

Suppose:
```text
Age = Attribute
```
Should contains a person's age.

Its domain might be something like 0-120, the domain defines what values are acceptable.

#### **5. Relation Schema**

A **relation schema** Describes the structure of a relation.

For example:
```text
STUDENT(Student_ID, Name, Age, Course_ID)
```

This tell us:
- Relation name = STUDETN
- Attributes = Student_ID, Name, Age, Course_ID

#### **6. Relation Instance**

The Actual collection of tuples at a particular point in time is called the **relation instance**.

**Schema vs Instance**

This distinction is extremely important.
| **Schema** | **Instance** |
|------------|--------------|
| Structure/Design | Actual data |
| Relatively stable | Changes fequently |
| Defines attributes | Contains tuples |
| Example: Student(ID, Name, Age) | Actual student records |

#### **7. Properties of a Relation**

A relational table follows certain principles.

**Property 1 - Each cell contains single values**

**Property 2 - No Duplicates Typles**

A relation is mathematically treated as a set of tuples, and sets do not contains duplicates.

This distinction is important:
> Relatinal theory treats realtions as sets; SQL tables can behave differently unless constraints are imposed.

**Property 3 - Order of rows doesn't matter**

**Property 4 - **Order of Attributes**

Similarly, relational theory identifies attributes by their names, not merely by their physical position.

#### **8. Integrity Constraints**

A relational database should not only stores data - it should prevent invalid data. These rules are called **integrity constraints**.

The major ones we should know are:
- _Domain integrity_: values must belongs to the allowed domain.

- _Identity integrity_: Primary keys can not be NULL.

- _Reffrential integrity_: A Foreign key should refer to a valid refrence value, subject to the database's foreign-key rules.


#### **9. Relational Shcema of a University Database**

Let's design a small database properly.

We might have:

```text
STUDENT (
    Student_ID,
    Name,
    Age,
    Course_ID
)

COURSE (
    Course_ID,
    Course_Name,
    Department_ID
)

DEPARTMENT (
    Department_ID,
    Department_Name
)
```

Keys:

```text
STUDENT
Student_ID -> Primary key
Course_ID -> Foreign key

COURSE
Course_ID -> Primary key
Department_ID -> Foreign key

DEPARTMENT
Department_ID -> Primary key
```

Relationships:
```text
DEPARTMENT 1 <---------> N COURSE 1 <-----------> N STUDENTS
```
This is the relation Database Schema.

#### **10. SQL Representation**

Now let's implement our Conceptual Model.

**Create the Database**

```SQL
CREATE DATABASE University;
```

**Department**

```SQL
CREATE TABLE Department (
    Dept_ID INT PRIMARY KEY,
    Dept_Name VARCHAR(100) NOT NULL
);
```

**Course**

```SQL
CREATE TABLE Course (
    Course_ID INT PRIMARY KEY,
    Course_Name VARCHAR(100) NOT NULL,
    Dept_ID INT,
    FOREIGN KEY (Dept_ID) REFERENCES Department(Dept_ID) 
);
```

**Student**

```SQL
CREATE TABLE Student (
    Student_ID INT PRIMARY KEY,
    Student_Name VARCHAR(100) NOT NULL,
    Age INT,
    Course_ID INT,
    FOREIGN KEY (Course_ID) REFERENCES Course(Course_ID)
);
```

#### **10. Retrieving Related Data**

Once the tables are connected, We can use SQL to combine information.

Suppose we want:
> Student name + course name + department name.

We can write.
```SQL
SELECT
    s.Name,
    c.Course_Name,
    d.Dept_Name
FROM Student as s
JOIN Course as c
    ON s.Course_ID = c.Course_ID
JOIN Department as d
    ON c.Dept_ID = d.Dept_ID;
```

This is where the relational model becomes incredibly powerful.

Instead of storing everything in one giant table, we keep related information separated and connect it when needed.

#### **11. Why not just Use One Giant Table?**

Imagine:
```text
Customer_ID
Customer_Name
Customer_City
Order_ID
Product_ID
Product_Name
Product_Price
Quantity
Payment_Method
Salesperson
...
```
This creates:
- Redundancy
- Update problems
- Insertion problems
- Deletion problems
- large storage requirements
- Inconsistent data.


#### **12. The Proper Definitions**

**"Explain the Relational Data Model."**
> The relational data model is a logical data model which organize data as relations, which commonly be known as tables consistinge rows and columns. Each realtion contains tupels representing records and attributes representing properties of those records. Attributes have defined domains of valid values. Keys such as primary keys and foreign keys are used to identify tuples and establish relationships between relations. The models also uses integrity constraints such domain integrity, identity integrity and referential integrity to maintain data accuracy and consistency. The relation model provides a structured and flexible foundation for storing, querying, and managing data in relational database systems.
