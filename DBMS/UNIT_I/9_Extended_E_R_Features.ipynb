## **Extended E-R Features(EER)**

### **1. Why do we need an Extended E-R Model?**

Let's start with a university database.

A university stores information about people. Some people are students, some are instructors, and some may be both.

Students have attributes such as `Student_ID` and `Program`.

Instructors have attributes such as `Instructor_ID` and `Salary`.

But both students and instructors share common attributes:
- Name
- DOB
- Addresses
- Phone number

How should we design this **database?**

One approach is to create separate tables with all their attributes repeated. But that duplicates unnecessarily and can form inconsistency.

Another approach is to recognize that **STUDENT** and **INSTRUCTOR** are specialized types of **PERSON**.

That's where EER becomes useful.

```text
                      PERSON (Person_ID, Name, DOB)
                                    | 
               |                                         |
    STUDENT(Student_ID, program)       INSTRUCTOR(Instructor_ID, salary)

```
**Simple Englis:** PERSON contains the common properties. STUDENT and INSTRUCTOR represent more specific types of person with additional properties.

### **2. Specialization**

**Specialization** is the process of dividing a general entity type into more specific subtypes based on different characteristics.

For example:
- PERSON -> STUDENT
- PERSON -> INSTRUCTOR

The general entity is called the **superclass**, and the more specific entities are called **subclasses**.

#### **Why is specialization useful?**

It lets us represent differences without repeatedly defining the same common information.

Imagine changing a person's address. If that person id represented in the **PERSON** superclass, we can update the shared address in one place rather than maintaining duplicate copies in multiple sybytpe records.

### **3. Generalization**
Now reverse the direction

Suppose we begin with two entity types:
- CAR: Registration number, Model, Number of seats
- TRUCK: Registration number, Model, Load capacity

We notice that both share properties such as registration number and model.

We can combine their common properties into a more general entity called **VEHICAL**.

**Generalization** combines multiple specific entity into a common super class.

### **4. Inheritance**

**Inheritance** means a subclass recieves the attributes and relationships of its superclass.

If PERSON has:

`Person_ID`, `Name`, `DOB`

And STUDENT has:

`Program`, `Admission_Date`

Then a student has access to the common PERSON properties as well as the student-specific ones.

### **5. Disjoints vs. Overlapping constraints**

**A. Disjoint specialization**

A PERSON can belong to **at most one** of its specified subclasses.

For example, if a college's rules classify that a PERSON can either be a STUDENT only or an INSTRUCTOR only, but never both, the subclasses are disjoint.

**B. Overlapping specialization**

A PERSON can belong to **more than one** subclass simultaneously.

For example,  a postgraduate student might also works as an instructor athe same university.

The key distinction:
- **Disjoint**: No entity instance can belong to more than one subclasses.
- **Overlapping**: An entity instance can belongn to multiple specified subclasses.

### **6. Total vs. Partial Specialization**

Now consider a different question.

Must every PERSON in the database be either a STUDENT or an INSTRUCTOR?

This determines whether specialization is total or partial.

- **Total specialization**: Every entity in the superclass must belongs to any of its specified subclasses.<br>**Example**: If the university's PERSON database contains only students and instructors, every PERSON must be one or both of these.

- **Partial specialization**: Some entities in the superclass may belong to none of the specified subclasses.<br>**Example**: the university also stores visitors or administrative personnel in PERSON, but those categories are not included in the STUDENT and INSTRUCTOR subclasses.

### **7. Specialization Hierarchy**

A subclass can be further divided into more specific subclasses.

For example:
STUDENT can be divided into GRADUATE and UNDERGRADUATE.
