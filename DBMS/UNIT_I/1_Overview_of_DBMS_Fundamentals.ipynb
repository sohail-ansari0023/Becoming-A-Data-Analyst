## **Overview of DBMS Fundamentals**

#### **1. What is Data?**
Data means raw facts or observations that have not yet been processed into meaninful information.

for example:

```text
101
Mohammad Sohail
commerce
66
Dhanbad
```

These are the individual facts. By themselves, they are simple **data**.

Data can be:

- Numbers
- Names
- Dates
- Age
- Addresses
- Marks
- Phone Numbers
- Place
- Customer Transactions
- Audio
- Vedio
- Images
- Text

**Why does data exist?**

Businesses and Orgranizations continously performs activities,

A School admits students, a bank records transactions, Amazon records Orders, and a hospital records patients.

All these activities produce data.

For example, When we buy a 500rs product online via any platform, those platforms may store:

```text
CustomerID = C1002
Product = Headphone
Unit = 1
Price = 500rs
OrderDate = 12-02-2026
Payment = UPI
```
That collection of facts is data.


#### **2. What is Information?**

Information is processed and organized data that have some meaning and help us to make decisions.

Suppose a company has this data:

```text
Product A -> Sales 100 units
Product B -> Sales 140 units
Product C -> Sales 80 units
```
After processing it, we figure out:

> Product B is the Best-Selling Poduct.

That coclusion is information.

So:

```text
Raw data -> Processing -> Information
```

**Why information is important?**

Because business doesn't collect data for the sake of just collecting it.

They collect them to find out answers to questions such as:

- Which product are the best-selling?
- Which region produces most revenues?
- Which employees performing their best?
- What kind of customers buys the most?
- Which month has the heighest sales?

And this is where the data becomes Useful.

#### **3. What is Database?**

A Database is an organized collection of related data that can be stored, accessed, managed and updates efficiently.

Imagine a college had stores information about 10,000 Students.

It might need to store:

```text
Student Id
Student Name
Date of Birth
Course
Semester
Year
Marks
Phone
Adress
```

instead of keeping all these in random files, we store them into Databases.

**Why do we need Databases?**

Because large organizations produce enormous amounts of data.

**A small shop**

100 customers -> Spreadsheets may be enough.

**Amazon**

Millios of customers + millions of orders + products + payments + deliveries and for that we need Database system.

The database provides organized storage and make retrieving information much easier.

> Database = Properly organized digital data storage.

#### **4. What is DBMS?**

DBMS stands for Database Management System, It is software system that enables users to create, store, update, delete and manage data from database, while providing facilities such as security, Data integrity, Concurrency control, backup and recovery.

Examples:
- MySQL
- PostgreSQL
- Oracle Database
- Microsoft SQL Server
- SQLite
- MariaDB

**Important Distinction**

A Database Stores data.

A DBMS is the software used to manage data.

Think of a library:

```text
Books            = Database
Librarian/System = DBMS
Student          = User
```

#### **5. Why was DBMS Needed?**

This is one the most important conceptual questions.

Before DBMS, Organizations heavily relied on **traditional file system**.

Suppose a college maintains separate file:

```text
Student file
Fees file
Exam file
Attendance file
```
This creates several problems.

1. Data Redundancy
2. Data Inconsistency
3. Difficulty in Data access.
4. Data Security
5. Data integrity
6. Data Sharing

**1. Data Redundancy**

Data Redundancy means Unnecessary Duplications of data.

Suppose:

Student ID 101:
```text
Student file -> Sohail
Fees file -> Sohail
Exam file -> Sohail
Attendance file -> Sohail
```

The same information may be repeated many times.
> Data Redundancy = Unnecessary duplication

**2. Data Inconsistency**

the system may contains conflicting informations.

_Why is this dangerous?_

Imagine a banking system:
```text
Customer database = 50,000 Rs
Another file = 45,000 Rs
```
Which one is corrct? That is a serious problem, and that is what Data inconsistency means.

**3. Difficulty in Data Access**

Imagin our manager asks:
> "Show me all customers from Dhanbad who purchased more than 50,000 Rs this year."

If data is stored in traditional files, answering this becomes difficult, since have to go through various files.

A _DataBase_ Allows us to use queries.

For example:
```SQL
SELECT *
FROM Customers
WHERE city = 'Dhanbad'
AND total_purchase > 50000;
```
The database can process this query and provide the required records.

**4. Data Security**

Not everyone should have access of every informations. A DBMS can control _Who can access what_. This is called database *security/access control*.

**5. Data integrity**

Data integrity means maintaining **accuracy, consistency, and reliability of data.**

Suppose a student's age is:
```text
Age = -25
```
Which is Obviously, Something wrong.

A DBMS can apply rules to prevent invalid data.

For example:
```SQL
Age INT CHECK (Age >= 0)
```
Similarly, if studen ID must be unique, the DBMS can enforce that rule.

**6. Data Sharing**

One of DBMS's biggest advantages is that multiple users and applications can work with the same databasse.

Imagine an online Shopping company.

At the same time:

```text
Customer
   ↓
Website

Warehouse
   ↓
Inventory System

Finance Department
   ↓
Payment System

Delivery Department
   ↓
Shipping System
```

all may interact with the same underlying data.

#### **6. What Does a DBMS Actually Do?**

Let's understand the DBMS as a complete system.

A DBMS generally allows us to:

```text
Create data
    ↓
Store data
    ↓
Retrieve data
    ↓
Update data
    ↓
Delete data
    ↓
Protect data
    ↓
Maintain consistency
    ↓
Backup / Recovery
```

A simple way to remember its role it:
> CRUD + Control

CRUD means:

C -> Create,  R -> Read, U -> Update, and D -> Delete

These are the four basic operations performed to a Data.

Suppose we have a student ID:
```text
Student_ID = 101
Name = Sohail
Course = Commerce
```
Create:
```SQL
INSERT INTO students VALUES (101, 'Sohail', 'Commerce');
```
READ:
```SQL
SELECT * FROM students;
```
Update:
```SQL
UPDATE students
SET course = 'B.com'
WHERE student_ID = 101;
```
Delete:
```SQL
DELETE FROM students
WHERE student_ID = 101;
```

#### **7. Different types of Database Users

Not everbody interacts with database in the same way.

**Database Administrator ---- DBA**

The _DBA_ is responsible for managing and controlling the Database environment.

Typical responsibilities include:
- Security
- Backup
- Recovery
- User permissions
- Performance
- Database maintenance

**Database Developer**

Developers builds applications that communicates with the database.

For example:
```text
Shopping Website
      |
Application Code
      |
Database
```
The users don't generally writes SQL directly.

The application may generates SQL behind the scene.

**Data Analyst**

A data analyst often Uses SQL to retrieve and analyse data.

Example:
```SQL
SELECT city, SUM(sales)
FROM orders
GROUP BY city;
```

> This tells Which city has generated the most sales?

So DBMS is not only a commerce topic ----it is a foundation for data analytics.

#### **8. What is a Database Management System Architecture?**

At a high level, think of the system like this:
```text
             USERS
               ↓
        APPLICATION / SQL
               ↓
             DBMS
        ┌──────┼──────┐
        ↓      ↓      ↓
     Security Query  Storage
        ↓    Processing Management
        └──────┼──────┘
               ↓
           DATABASE
```

The users doesn't normally cares how the data has been stored in database.

The DBMS acts as a **midle layer**.

This is one of its most important role.


#### **Importand Terminology**

Before we go deeper into DBMS, we should get comfortable with these words.

**Database**

Organized collection of related data.

**DBMS**

Software used to manage databases.

**Table**

A structured collection of data arranged in rows and columns.

Example:
| Student_ID | Name | Subject | Semester | Year |
|----------- |------|---------|----------|------|
| 101 | Sohail | Commerce | 7 | 2026 |

**Row / Record / Tuple**

One complete entry in a table.

Example:
```text
| 101 | Sohail | Commerce | 7 | 2026 |
```

**Columns / Attribute / Field**

A particular property of a data.

Example:
```text
| Student_*ID | Name | Subject | Semester | Year |
```

**Query**

A request to retrieve or manipulate data.

Example:
```SQL
SELECT * From students;
```

**SQL**

Structured Query Language used to interact with relational databases.
