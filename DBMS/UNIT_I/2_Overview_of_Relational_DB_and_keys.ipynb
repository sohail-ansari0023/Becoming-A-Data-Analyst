## **Overview of Relational Databases and Keys**

#### **1. What is relational Databases?**

A relational Database is a database that stores data in the form of **tables**, where tables are connected to each other through relaitonships.

A table consists of:
- **Row** -> individual records
- **Columns** -> attributes/properties.

For example:

**Students**
| **Student_ID** |**Name** | **Course** | **City** |
|----------------|---------|------------|----------|
| 101 | Sohail | Commerce | Dhanbad |
| 102 | Rohit  | BA       | Ranchi  |
| 103 | Ayesha | B.Sc     | Dumka   |

**Courses**
| **Course_ID** | **Course_Name** | **Duration** |
|---------------|-----------------|--------------|
| C01 | commerce | 4 Years |
| C02 | BA       | 4 Years |
| C03 | B.Sc     | 4 Years |

The important point is that these tables can be related. And that's why it is called **relational Database**.

#### **2. Why is it called relational**

The word relation has a specific meaning in relational Databases theory.

In practical term, relation mean a table.

For example:
> Table(students) is a relation, Table(courses) is another relation.

They can be connected through a common attribute.

```text
Students <-------------> Courses
            Course_ID
```

This connection represents a relationship between the data.

#### **4. What is a Key?**

The heart of this topic....

**Simple Definition**

A key is an attribute or set of attributes used to **identify records uniquely or stablish relationships between the tables**.

Suppose:
| **Student_ID** | **Name** | **City** |
|----------------|----------|----------|
| 101 | Sohail | Dhanbad |
| 102 | Rohit  | Ranchi  |
| 103 | Ayesha | Dumka   |

Could we Identify a studen with `City`?

No, Because multiple students may from the same city.

Cound we use `Name`?

No again, Multiple Students can have same name.

But:
```text
Student_ID
```
Can be used to uniquely identify each student. Therefore:
> Student_ID is a key candidate.

#### **5. Types of Keys**

The major keys we needs to understand are:
- Super Key
- Candidate key
- Primary key
- Foreign key
- Alternate key
- Composite key
- Unique key
- Natural key
- Sauragate key

**=> Super key**<br>

A _super key_ is an attribute or combinations of attributes used to uniquely identify a row.

**=> Candidate Key**<br>

A _Candidate key_ is the minimal _super key_.

The word minimal is extreme important.

> We cannot remove an attribute from the key and still uniquely identify the record.

Suppose:
```text
Student_ID is unique.
Email is also unique.
```
Then both can be candidate keys. However why we aren't calliing `student_ID` + `Email` a candidate key? Because we don't need both. `student_ID` is enought to identify the rows uniquely.

Therefore:
```text
Super key
     ↓
Minimal Super key
     ↓
Candidate key
```

**=> Primary key**<br>

One of the candidate keys Chosen as the main identifier. The selected candidate key becomes the: `Primary Key`.

Important properties of Primary keys:
- Must uniquely identify each row.
- Cannot contains any `NULL`
- Should remain stable
- Is normally chosen for candidate keys

SQL
```SQL
CREATE TABLE students (
    Student_ID VARCHAR(10) PRIMARY KEY,
    Name VARCHAR(100),
    Email VARCHAR(100)
);
```
**=> Foreign key**

A _Forign key_ is an attribute in one table that refers to key (Typically the Primary Key) of another table.

**=> Alternate key**

Those candidate keys which aren't selected as Primary key, called Alternate key.

**=> Composite key**

Sometimes one columns isn't enough to uniquely identify each records. The we can use multiple columns together. To form a key, that keys are called composite key.

Example:
> Student_ID + Email 

**=> Unique key**

A _UNIQUE_ constraint ensures that values in a column or set of columns do not duplicate.

For example:
```SQL
CREATE TABLE students(
    Student_ID INT PRIMARY KEY,
    Name VARCHAR(100),
    Email VARCHAR(100) UNIQUE
)
```

**Natural Key vs Surrogate Key**

- _Natural key_: A natural key is a key that contains actual meaning in real worlds.
    - Examples: Email, Passport Number, ISBN.

- _Surrogate key_: A Surrogate key is generally created as artificial identifier with no business meaning.
    -  Example: Student_Id, Course_ID 

**Why use surrogate keys?**

- simple
- short
- stable
- easy to reference

#### **6. Relationships between tables**

Rlational database commonly represents three basic relationship types.

**1. One-to-One (1:1)**

One record in table A corresponds to one record in table B.

**2. One-to_Many (1:N)**

One record in table A relates to Many record in table B.

Example:
```text
Course -> Students i.e Many students can have one courses.
```

**3. Many-to-Many (M:N)**

Many records in A relates to Many records in table B.

Example:
```text
Students -> Courses i.e A student can take mulitple courses.
and a course can have multiple students.
```

#### **The most important concept is**

Primary key = Identifies records

Foreign key = establishes relationships between tables

## **Exam Oriented Definitions**

**Relational Database**
> A relational database is database system that organize data into the form of tables consisting rows and columns along with keys which establishes relaionships amongst multiple tables.

**Primary key**
> A primary key is an attribute or a set of attributes that is used to uniquely identify each record in a table which cannot contains NULL values.

**Foreign key**
> A Freign key is an attribute or set of attributes in one table that refferences a key, usually a primary key, of another table to establish relationship between the tables.

**Candidate key**
> A candidate key is minimal super key that can uniquely identify each record in a relation.

**Super key**
> A super key is an attribute or combinations of attributes that can uniquely identify each record in a relation.

**Composite key**
> A composite is a key that consists of two or more attributes used together to uniquely identify records.
