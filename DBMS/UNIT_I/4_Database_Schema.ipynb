## **Database Schema and Formal Relational Query language**

### **Part A — Database Schema**

#### **1. What is a Database schema?**

A Database schema is the logical blueprint or structure of a database. It defines:
- What table exist,
- What kind of attributes does the table have,
- data types,
- relationships between tables,
- keys,
- Constraints, and
- Other structural rules.

In Simple words:
> A database schema told us how the database is designed, not what data is currently stored in it.

For example:
```text
STUDENT(Student_ID, Name, Age, Course_ID)

COURSE(Course_ID, Course_Name, Department_ID)

DEPARTMENT(Department_ID, Department_Name)
```

A well designed schema provides structural rules of the data base, to prevent data inconsistency.

#### **2. Database Schema as a Formal Definition**

A relational database schema is the collection of relation schemas. Each realtion schema specifies:
- Relation name,
- Attributes,
- Domains/Data types
- constraints

### **Part B — Formal Relational Query Languages**

#### **1. What is a Query Language?**

A **Query Language** is a language that is to request information from the database and, depending on this language we perform operations on database data.

For example:
```SQL
SELECT *
FROM students;
```
means:
> Give all records of students relation/table.

SQL is the practical query language we will all gonna use.

But before SQL, relational database theory developed **formal query languages** based on mathematical concepts.

The two major formal query languages are:
- **Relational Algebra**
- **Relational Calculus**

**SQL** is more practical and user-friendly, but relational algebra and calculas provide important theoratical foundations.

#### **2. Relational Algebra**

Relational Algebra is a procedural formal query language that uses a collection of operations to manipulate relations and produce new relations.

The key word is:
> **Procedural** -> What operations should be performed to obtain the result?

For example:
```text
Take STUDENT
      ↓
Select Commerce students
      ↓
Project their names
      ↓
Return result
```

Relaional algebra takes relations as an input and produces a relations as output.

#### **3. Basic Algebra Operations:**

The fundamental operations are:
- Selection
- Projection
- Union
- Set difference
- Cartesian product
- Rename
- Intersection
- Joins
- Division

Basically most of them are derived from **mathematical set theory.**

#### **4. Relational Algebra Query Example**

Let's solve a small business problem.

_Suppose we have:_
```text
CUSTOMER(Customer_ID, Name, City)
```
_and:_
```text
ORDER(Order_ID, Customer_ID, Amount)
```
Question:

>Find the names of customers from dhanbad

The SQL Equivalent:
```SQL
SELECT Name
FROM CUSTOMER
WHERE City = 'Dhanbad';
```

This is a beautiful example of how SQL concepts connect to relational algebra.

#### **5. Relational Calculas**

The second formal query language.

**Definition**

**Relational Calculas** is a non-procedural formal query language that specifies what data is required rather than how to obtain it.

The key difference.

Relational Algebra:
> How should we obtain it?

Relational Calculas:
> What data do we want?

#### **6. Procedural vs Non-Procedural**

Suppoes we ask:
> Fund students whose marks are greater than 70.

**Relational Algebra**

we specify an operation:
```text
σ Marks > 70 (STUDENT)
```
we're describing an operation to obtain the result.

**Relational Calculus**

We describe the desired tuples:
> Give me student tuples for which marks > 70.

our focus on the **condition defining the result**, not the sequence of operations.

#### **7. Types of Relational calculus**

There are two major forms:

**1. Tuple Relational Calculus — TRC**

It uses _tuples/rows_ as variables.

General from:
```text
{ t | P(t) }
```
Meaning:
> Find all tuples `t` for which predicate `P(t)` is true.

For example, conceptually:
```text
{ t | t ∈ STUDENT AND t.Course = 'Commerce' }
```
means:
> Find student tuples whose course is Commerce.

**2. Domain Relational Calculus — DRC**

It uses _individual attribute values_ rather than complete tuples as variables.

General Form:

```text
{ <x₁, x₂, ..., xₙ> | P(x₁, x₂, ..., xₙ) }
```
It describes the required values from the domains of the attributes.

### **Relational Algebra and SQL**

Now connect theory with practical SQL.

### **A More Advanced Idea - Query Optimization**

Here's the theory becomes useful in real database systems.

Suppose we need:
```text
Customers from Dhanbad
who placed orders above 50,000.
```
One conceptual approach could be:
```text
CUSTOMER × ORDER
        ↓
Huge intermediate result
        ↓
Filter
```
That's potentially expensive.

A database optimizer may instead push filtering earlier:
```text
CUSTOMER
   ↓
Select Dhanbad
   ↓
Smaller relation
   ↓
Join with orders
   ↓
Filter/order calculations
```

Why?

Because processing smaller amount of data is often cheaper. 

This is one reason realtional algebra is not merely academic theory.

It provides a foundation for reasoning about **query execution and optimization**.
