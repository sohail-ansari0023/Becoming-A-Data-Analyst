Now we need to answer a deeper question:
> **What rules control how entities are allowed to participate in relationships?**

Those rules are called **E-R constraints**.

### **1. What Are E-R Constraints?**

An _E-R constraint_ is the rule which that restricts how enteties can participate in relationships.

Why do we need restrictions?

Consider:
```text
STUDENT -- enrolls in -- COURSE
```
Withou any rules, we don't know:
- Can one student enroll in one course or many?
- Can one course have one student or many?
- Must every student enroll in a course?
- Can a course exist without students?
- Can a student enroll in zero courses?
- Can an entity pariticipate multiple times?

Constraints answer these questions.

So:
> **E-R Constraints make the conceptual medel reflect the rules of the real-world system**.

### **2. Main Types of E-R Constraints**

For this syllabus, we've to focus on these:
```text
1. Key Constraints
2. Cardinality Constraints
3. Participation Constraints
```

#### **1. Key constriants**

First:
> A key identifies an entity uniquely.

But in the context of E-R relationship, a **key constraints** helps specify how many entities of one entity set can be associated with an entity of another set.

Let's see why....

Suppose:
```text
EMPLOYEE --- Works_for ---- DEPARTMENT
```
Assume the rule is:
> An employee can work for only one department.

But:
> A department can have many employees.

Then:
```text
EMPLOYEE N ---- 1 DEPARTMENT
```
or equivalently:
```text
DEPARTMENT 1 ---- N EMPLOYEE
```
The 'one' side tells us that a particular employee can be associated with **at most one department**.

This is a key contraint on the relationship.

#### **2. Cardinality Constraints**

cardinality describes the possible number of relationship instances.

The common cardinalities are:
```text
1 : 1
1 : N
N : 1
M : N
```
We've already completed them, let's skip here.

### **Cardinality isn't the whole story**

Here's where many students make a mistake.

Suppose:
```text
DEPARTMENT 1 ----- N EMPLOYEE
```
We know:
> One department can have many employees.

But another question remains:
> Must every department have at least one employee?

Cardinality alone doesn't completely answer these questions. This is where **participation constraints** come in.


#### **3. Participation Constraints**

Participation constraints tells us wheter the existence of an entity requires it to participate in a relationship.

There are two main types:
```text
1. Total Participation
2. Partial Participation
```

#### **1. Total Participation**
An entity set has toatl participation in a relationship if **every entity in that entity set must participate in at least one relationship instance.**

Example:
```text
EMPLOYEE ---- Works for ---- DEPARTMENT
```
Suppose the business rule says:
> Every employee must work for a department.

Then every employee must participate in `work_for`

Therefore:
```text
EMPLOYEE
    ↓
Total participation
```
There cannot be:
```text
Employee 101 -> No department
```
under this rule.

#### **2. Partial Participation**

An entity set has _partial participation_ if some entities may participate in the relationship, while others may not.

Example:
```text
EMPLOYEE --- manages --- DEPARTMENT
```
Suppose:
> Not every employee is a manager.

Then:
```text
Employee 101 → manages Department A
Employee 102 → no department managed
Employee 103 → no department managed
Employee 104 → manages Department B
```
Some employees participate, some don't.

Therefore: **Employee has partial participation in `manages`.**

#### **Total vs Partial Participation**

| **Total Participation** | **Partial Participation**|
|-------------------------|--------------------------|
| Participation is mendotary | Participation is optional|
| Every entity participate | some entities may not participate |
| Minimum participation = 1 | Minimum Participation = 0 |
| Example: Every employeee must belong to a department | Example: not every employee manages a department |

This "minimum" idea is extremely useful.

Think:
```text
Total -> minimum 1
Partial -> minimum 0
```

### **Combining Cardinality and Participation**

No we're getting to the really useful part.

Cardinality tells us:
> **Maximum?**

Participation tells us:
> **Minimum?**

This is very useful mental model.

For example:
```text
EMPLOYEE ---- Works for ---- DEPARTMENT
```
Suppose:
- Every employee must belongs to exactly one department
- A department can have zero or many employees.

then we can express:
```text
EMPLOYEE:   1..1
DEPARTMENT: 0..N
```
Meaning:

For an employee:
```text
Minimum = 1
Maximum = 1
```
For a department:
```text
Minimum = 0
Maximum = many
```
This is much more precise than simply saying "1:N".

### **Minimu and Maximum Cardinality**

We can think of paticipation/cardinality together as:

```text
(minimum, maximum)
```
Examples:
- (0,1) = zero or one
- (1,1) = Exactly one
- (0,N) = zero or many
- (1,N) = one or many

This notation is common in variou E-R and UML representations.

#### **Example: University Database**

Let's take:
```text
STUDENT ----- Enorlls_In ------ COURSE
```

Suppose a university rules are:
1. A student may enroll in several courses.
2. A student may temporarily have no enrollment.
3. Every course must have at least one student.
4. A course can have many students.

Then:

**Student side**
```text
(0, N)
```
A student can have zero or many courses.

**Course side**
```text
(1, N)
```
A course must have at least one student and can have many.

So:
```text
STUDENT (0, N) -----Enrolls_In------(1, N) COURSE
```
This communicates much more information.


### **How These Constraints Affect Relational Tables**
Now let's connect this back to SQL.

Suppose:
```text
CUSTOMER (0, N) ------ Places ------ (1, 1) ORDERS
```
We could create:
```SQL
CREATE TABLE Customer(
    customer_id INT PRIMARY KEY,
    name VARCHAR(40)
);
```
Then:
```SQL
CREATE TABLE Order(
    order_id INT PRIMARY KEY,
    order_date DATE,
    customer_id INT NOT NULL,
    FOREIGN KEY (ccustomer_id)
        REFERENCES Customer(customer_id)
);
```

Notice:
```SQL
customer_id INT NOT NULL,
```
Why `NOT NULL`?

Because every order **must** have a customer.

That's the relational implementation of the mendotary participation idea.

And:
```SQL
FOREIGN KEY (customer_id)
```
ensures the customer referenced by an order must exist, subject to the FK rules.

### **Constraints Reading Practice**

Let's practice interpreting requirements.

**Examples:**
> "Each employee works for exactly one department".

Interpretation: `(1, 1)` mini = 1 and max = 1

> "A department can employ many employee".

Interpretation: `(0, N)` mini = 0 and max = N 

> "Every course must belong to a department."

Interpretation: `(1, 1)` mini = 1 and max = 1 (assuming exactly one department).

> "A student may enroll in several other courses."

Interpretation: `(0, N)` mini = 0 and max = N

> "Every course must have at least one student."

Interpretation: `(1, N)` mini = 1 and max = N


### **The proper definition**

**"Explain E-R constraints"**

> E-R constraints are rules used in the Entity-Relationship model to restrict the manner in which entities participate in relationships. They ensure that the conceptual database design correctly represents real-world business rules. Important E-R constraints include key constraints, cardinality constrints, and participation constraints. Cardinality specifies the maximum number of relationship instances in which an entity can participate, shuch as 1:1, 1:N, and M:N. Participation specifies wheter the participation is mandetory or optional and can be total or partial. These constraints help ensure consistency and guide the conversion of an E-R model into relational schemas.
