# **This notebook covers Python Timedelta Object from the built-in `datetime` module**

Understanding how to manipulates time is really necessary for handling time Series datasets or engineering temporal features for productive models.

In Python, the `timedelta` object represents a duration - difference between two dates or times.

### **1. Creating timedelta**

A `timedelta` object is created by passing specific keywords arguments. All arguments are optional and defaults to 0.

```python
from datetime import timedelta

# Signature.
timedelta(days=0, seconds=0, microseconds=0, milliseconds=0, minutes=0, hours=0, weeks=0)

# Example: 2 weeks , 3 days, and 4 hours
date = timedelta(weeks=2, days=3, hours=4)
```

**NOTE**: _Internally, Python normalizes everything in `days`, `seconds`, and `milliseconds`. if we pass `weeks=1`. Pthon stores it as `days=7`._


### **2. The Complete Documentation(Attributes & Methods)**

Here is everything the `timedelta` class offers.

**Instance Attributes**
| Attribute | Description |
|-----------|-------------|
| `days`    | between -999,999,999 and 999,999,999 inclusive.|
| `seconds` | between 0 and 86,399 inclusive(number of seconds in a day)|
| `microseconda` | between 0 and 999,999 inclusive.|

**Class Attributes**
| Attribut | Description |
|----------|-------------|
| `timedelta.min` | The most negative timedelta object (`timedelt(-999,999,999)`) |
| `timedelta.max` | The most Positive timedelta object |
| `timedelta.resolution` | The smallest possible difference between non-equal timedelta objects(`timedelta(microseconds=1)`) |

**Instance Methods**
| Method | Description |
|--------|-------------|
| `total_second()` | Returns the total duration contained in the object. Expressed purely in seconds(float). |

**Supported Operations**<br>
we can perform standard Arithematic operations with timedeltas and `datetime` objects.
- *Additions/Substraction:* `datetime` + `timedelta` (shifts a date forward/backward).

- *Delta Math:* `timedelta` + `timedelta` or `timedelta` - `timedelta`.

- *Multiplication/Division:* `timedelta * float` or `timedelta / timedelta` (yields a float ratio.)

- *Comparisons:* `<`, `>`, `<=`, `>=`, `==`.

### **3. The Absolute Necessary Cut**
When building industry-level data pipeline or cleaning datasets, we won't need most of the class attributes (`min`, `max`, `resolution`). we really only need to master these 3 core concepts:

**A: Extracting total Seconds for Feature Engineering.**

`total_seconds()` is the bridge to converts a time duration into a clean, continous numerical features.

```python
from datetime import datetime

starting_date_sample: datetime = datetime(year=2025, month=8, day=5, hour=3, minute=35, second=0)
# just increased day=5 to day=6, which gives exactly how much seconds in a day.
ending_date_sample: datetime = datetime(year=2025, month=8, day=6, hour=3, minute=35, second=0)

# Calculating the durations
durations: datetime = ending_date_sample - starting_date_sample

# converting to pure numerical features: (float).
num_of_seconds: float = durations.total_seconds()
num_of_seconds # Output: 86400.0 (Total number of seconds in a day).
```

**B. Time Shifting (If we wanna shift the date or time or year or anything)**

Whether we want to define rolling period(timing) (20 days ago, or 24 hours later). we can achieve it by adding and substracting `timedelta` object from a specific time period.

```python
from datetime import datetime, timedelta

# current date and time.
current_data: datetime = datetime.now()

# let's go 20 days ago.
cutoff_date: datetime = current_data - timedelta(days=20)
cutoff_date
```

**C. Division ratios**

some time we need to know how many times a specific duration fits into a specific time period. like:

- _How many weeks are present in 150 days._


```python
from datetime import timedelta

# 1. Total Time (150 days).
total_time_1: datetime = timedelta(days=150)

conversion_time: datetime = timedelta(weeks=1)

# just divide the conversion time with the total time.
converted: datetime = total_time_1/conversion_time
print(f"There are total {converted:.2f} weeks in {total_time_1.days} days.")
```
